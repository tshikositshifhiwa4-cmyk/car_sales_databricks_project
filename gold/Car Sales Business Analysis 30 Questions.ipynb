# Car Sales Business Analysis

## Overview
This notebook contains SQL queries to answer 30 key business questions using the `practice.bricks.car_sales_analytical_view` table.

## Question Categories:

### Revenue & Sales Volume (Q1-3)
* Total revenue, vehicles sold, average price

### Time Trends (Q4, Q16-18)
* Monthly/quarterly/yearly trends, weekday vs weekend patterns

### Vehicle Analysis (Q5-9, Q25-27)
* Top makes, models, body styles, features, and combinations

### Geographic Analysis (Q10-12)
* Performance by state, above-market locations

### Seller Performance (Q13-15)
* Top sellers by revenue, volume, and market premium

### Market Value Analysis (Q19-21)
* MMR comparisons, which vehicles/sellers outperform expectations

### Price Factors (Q22-24)
* Impact of mileage, condition, value retention

### Strategic Insights (Q28-30)
* Key price drivers, undervalued/overvalued vehicles, profitability opportunities

---
**Total Records:** 550,284 sales transactions  
**Data Source:** practice.bricks.car_sales_analytical_view

```sql
-- Q1: Total revenue from car sales
-- Q2: Total vehicles sold
-- Q3: Average selling price per vehicle

SELECT 
  SUM(selling_price) AS total_revenue,
  COUNT(*) AS total_vehicles_sold,
  AVG(selling_price) AS avg_selling_price
FROM practice.bricks.car_sales_analytical_view;
```

```sql
-- Q4: Are sales increasing or decreasing over time?

SELECT 
  sale_year,
  sale_month,
  sale_month_name,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS monthly_revenue,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY sale_year, sale_month, sale_month_name
ORDER BY sale_year, sale_month;
```

```sql
-- Q5: Which car makes sell the most units?
SELECT 
  make,
  COUNT(*) AS units_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY make
ORDER BY units_sold DESC
LIMIT 20;

-- Q6: Which car models generate the highest revenue?
SELECT 
  make,
  model,
  COUNT(*) AS units_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY make, model
ORDER BY total_revenue DESC
LIMIT 20;

-- Q7: Which trims or body styles are most profitable?
SELECT 
  body_style,
  COUNT(*) AS units_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price,
  SUM(selling_price - mmr) AS total_profit_vs_mmr
FROM practice.bricks.car_sales_analytical_view
GROUP BY body_style
ORDER BY total_profit_vs_mmr DESC;

-- Q8: Do automatic or manual vehicles sell for higher prices?
SELECT 
  transmission,
  COUNT(*) AS units_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_premium
FROM practice.bricks.car_sales_analytical_view
GROUP BY transmission
ORDER BY avg_price DESC;

-- Q9: Which colors or interiors are most common or valuable?
SELECT 
  color,
  interior,
  COUNT(*) AS units_sold,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY color, interior
ORDER BY units_sold DESC
LIMIT 20;
```

```sql
-- Q10: Which states generate the highest sales revenue?
SELECT 
  state,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY state
ORDER BY total_revenue DESC;

-- Q11: Which states sell the most vehicles?
SELECT 
  state,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue
FROM practice.bricks.car_sales_analytical_view
GROUP BY state
ORDER BY vehicles_sold DESC;

-- Q12: Are certain locations selling vehicles above market value?
SELECT 
  state,
  COUNT(*) AS vehicles_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_premium,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY state
ORDER BY avg_premium_pct DESC;
```

```sql
-- Q13: Which sellers generate the most revenue?
SELECT 
  seller_name,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY seller_name
ORDER BY total_revenue DESC
LIMIT 20;

-- Q14: Which sellers sell the highest number of vehicles?
SELECT 
  seller_name,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue
FROM practice.bricks.car_sales_analytical_view
GROUP BY seller_name
ORDER BY vehicles_sold DESC
LIMIT 20;

-- Q15: Which sellers consistently sell above MMR?
SELECT 
  seller_name,
  COUNT(*) AS vehicles_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_premium,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct,
  SUM(CASE WHEN selling_price > mmr THEN 1 ELSE 0 END) AS above_mmr_count,
  SUM(CASE WHEN selling_price > mmr THEN 1 ELSE 0 END) / COUNT(*) * 100 AS above_mmr_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY seller_name
HAVING COUNT(*) >= 10
ORDER BY avg_premium_pct DESC
LIMIT 20;
```

```sql
-- Q16: Which months or quarters perform best?
-- Q17: Weekday vs weekend trends
-- Q18: Which year had the strongest performance?

-- Monthly performance
SELECT 
  sale_month_name,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY sale_month_name, sale_month
ORDER BY sale_month;

-- Quarterly performance
SELECT 
  sale_year,
  QUARTER(sale_date) AS quarter,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue
FROM practice.bricks.car_sales_analytical_view
GROUP BY sale_year, QUARTER(sale_date)
ORDER BY sale_year, quarter;

-- Weekday vs weekend
SELECT 
  sale_weekday,
  CASE 
    WHEN sale_weekday IN ('Saturday', 'Sunday') THEN 'Weekend'
    ELSE 'Weekday'
  END AS day_type,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue
FROM practice.bricks.car_sales_analytical_view
GROUP BY sale_weekday
ORDER BY 
  CASE sale_weekday
    WHEN 'Monday' THEN 1
    WHEN 'Tuesday' THEN 2
    WHEN 'Wednesday' THEN 3
    WHEN 'Thursday' THEN 4
    WHEN 'Friday' THEN 5
    WHEN 'Saturday' THEN 6
    WHEN 'Sunday' THEN 7
  END;

-- Yearly performance
SELECT 
  sale_year,
  COUNT(*) AS vehicles_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price
FROM practice.bricks.car_sales_analytical_view
GROUP BY sale_year
ORDER BY sale_year;
```

```sql
-- Q19: Are vehicles selling above or below MMR?
SELECT 
  COUNT(*) AS total_vehicles,
  SUM(CASE WHEN selling_price > mmr THEN 1 ELSE 0 END) AS above_mmr_count,
  SUM(CASE WHEN selling_price <= mmr THEN 1 ELSE 0 END) AS below_mmr_count,
  SUM(CASE WHEN selling_price > mmr THEN 1 ELSE 0 END) / COUNT(*) * 100 AS pct_above_mmr,
  AVG(selling_price - mmr) AS avg_premium,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0;

-- Q20: Which makes or models outperform market expectations?
SELECT 
  make,
  model,
  COUNT(*) AS units_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_premium,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct,
  SUM(CASE WHEN selling_price > mmr THEN 1 ELSE 0 END) / COUNT(*) * 100 AS pct_above_mmr
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY make, model
HAVING COUNT(*) >= 20
ORDER BY avg_premium_pct DESC
LIMIT 20;

-- Q21: Which sellers or locations achieve the best market premiums?
-- By Seller
SELECT 
  seller_name,
  COUNT(*) AS vehicles_sold,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY seller_name
HAVING COUNT(*) >= 10
ORDER BY avg_premium_pct DESC
LIMIT 20;

-- By Location
SELECT 
  state,
  COUNT(*) AS vehicles_sold,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY state
ORDER BY avg_premium_pct DESC;
```

```sql
-- Q22: How does odometer mileage affect selling price?
SELECT 
  CASE 
    WHEN odometer < 50000 THEN '0-50k'
    WHEN odometer < 100000 THEN '50k-100k'
    WHEN odometer < 150000 THEN '100k-150k'
    WHEN odometer < 200000 THEN '150k-200k'
    ELSE '200k+'
  END AS mileage_range,
  COUNT(*) AS vehicles_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr
FROM practice.bricks.car_sales_analytical_view
GROUP BY 
  CASE 
    WHEN odometer < 50000 THEN '0-50k'
    WHEN odometer < 100000 THEN '50k-100k'
    WHEN odometer < 150000 THEN '100k-150k'
    WHEN odometer < 200000 THEN '150k-200k'
    ELSE '200k+'
  END
ORDER BY 
  CASE 
    WHEN mileage_range = '0-50k' THEN 1
    WHEN mileage_range = '50k-100k' THEN 2
    WHEN mileage_range = '100k-150k' THEN 3
    WHEN mileage_range = '150k-200k' THEN 4
    ELSE 5
  END;

-- Q23: Does vehicle condition significantly impact selling price?
SELECT 
  condition,
  COUNT(*) AS vehicles_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_premium
FROM practice.bricks.car_sales_analytical_view
GROUP BY condition
ORDER BY CAST(condition AS INT);

-- Q24: Which vehicle categories retain value best despite mileage?
SELECT 
  make,
  body_style,
  AVG(odometer) AS avg_mileage,
  AVG(selling_price) AS avg_price,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct,
  COUNT(*) AS units_sold
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0 AND odometer > 100000
GROUP BY make, body_style
HAVING COUNT(*) >= 10
ORDER BY avg_premium_pct DESC
LIMIT 20;
```

```sql
-- Q25: Which body styles are most in demand?
SELECT 
  body_style,
  COUNT(*) AS units_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY body_style
ORDER BY units_sold DESC;

-- Q26: Are newer model years consistently more profitable?
SELECT 
  vehicle_year,
  COUNT(*) AS units_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_premium,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY vehicle_year
ORDER BY vehicle_year DESC;

-- Q27: What vehicle combinations (make + model + trim) perform best?
SELECT 
  make,
  model,
  trim,
  COUNT(*) AS units_sold,
  SUM(selling_price) AS total_revenue,
  AVG(selling_price) AS avg_price,
  AVG((selling_price - mmr) / mmr * 100) AS avg_premium_pct
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY make, model, trim
HAVING COUNT(*) >= 5
ORDER BY total_revenue DESC
LIMIT 30;
```

```sql
-- Q28: What factors most strongly influence selling price?
-- Correlation analysis with key factors
SELECT 
  'Vehicle Year' AS factor,
  CORR(vehicle_year, selling_price) AS correlation
FROM practice.bricks.car_sales_analytical_view
UNION ALL
SELECT 
  'Odometer',
  CORR(odometer, selling_price)
FROM practice.bricks.car_sales_analytical_view
UNION ALL
SELECT 
  'MMR',
  CORR(mmr, selling_price)
FROM practice.bricks.car_sales_analytical_view
UNION ALL
SELECT 
  'Condition',
  CORR(CAST(condition AS DOUBLE), selling_price)
FROM practice.bricks.car_sales_analytical_view;

-- Q29: Which vehicles are undervalued or overvalued relative to MMR?
-- Undervalued (selling below MMR significantly)
SELECT 
  'Undervalued' AS category,
  make,
  model,
  vehicle_year,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(mmr - selling_price) AS avg_undervalue,
  COUNT(*) AS units
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0 AND selling_price < mmr * 0.8
GROUP BY make, model, vehicle_year
HAVING COUNT(*) >= 3
ORDER BY avg_undervalue DESC
LIMIT 10;

-- Overvalued (selling above MMR significantly)
SELECT 
  'Overvalued' AS category,
  make,
  model,
  vehicle_year,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_overvalue,
  COUNT(*) AS units
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0 AND selling_price > mmr * 1.2
GROUP BY make, model, vehicle_year
HAVING COUNT(*) >= 3
ORDER BY avg_overvalue DESC
LIMIT 10;

-- Q30: Where can the business maximize profitability?
SELECT 
  state,
  make,
  body_style,
  COUNT(*) AS units_sold,
  AVG(selling_price) AS avg_price,
  AVG(mmr) AS avg_mmr,
  AVG(selling_price - mmr) AS avg_profit_per_unit,
  SUM(selling_price - mmr) AS total_profit
FROM practice.bricks.car_sales_analytical_view
WHERE mmr > 0
GROUP BY state, make, body_style
HAVING COUNT(*) >= 5
ORDER BY total_profit DESC
LIMIT 30;
```
